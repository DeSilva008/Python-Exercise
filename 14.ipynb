#### Pandas

```python
import pandas as pd
import numpy as np
```

Series

```python
#Creating series with default index

num = [1,2,3,4,5]

nw_series = pd.Series(num)

print(nw_series)
```

```python
#Creating series with custom index

nw_series = pd.Series([1,2,3,4,5], index=[1,2,3,4,5])

print(nw_series)
```

```python
fruits = ['apple', 'papol', 'kehel']

fr_series = pd.Series(fruits, index=[1,2,3])
print(fr_series)
```

```python
#Creating series from a dictionary
dct = {'name':['ksaun', 'kamal', 'nimal'],
       'country':'Finland',
       'city':'Helsinki'
       }

seris_fr_dicit = pd.Series(dct)
print(seris_fr_dicit)
```

```python
#Creating a pandas series using numpy linespace

nw_series = pd.Series(np.linspace(2,10, num=10, endpoint=False)) #endpoint decide whether to add the end number to the array

print(nw_series)
```

Data Frames

```python
#Creating a dataframe from list of lists

data = [
    ['Asabeneh', 'Finland', 'Helsink'],
    ['David', 'UK', 'London'],
    ['John', 'Sweden', 'Stockholm']
]

df = pd.DataFrame(data)

print('without coloumn labeling:\n\n',df, '\n')

df = pd.DataFrame(data, columns=['Name', 'Country', 'City'])

print('With column labeling:\n\n', df)
```

```python
#Creating dataframe usind dictionary

data = {'Name': ['Asabeneh', 'David', 'John'],
        'Country':['Finland', 'UK', 'Sweden'], 
        'City': ['Helsiki', 'London', 'Stockholm']
        }

df = pd.DataFrame(data)
# print(df)
df
```

```python
#Creating a dataframe from list of dictionaries
data = [
    {'Name': 'Asabeneh', 'Country': 'Finland', 'City': 'Helsinki'},
    {'Name': 'David', 'Country': 'UK', 'City': 'London'},
    {'Name': 'John', 'Country': 'Sweden', 'City': 'Stockholm'},
    {'Name': 'Mark', 'Last Name': 'Casthro', 'City': 'LA'}
    ]

df = pd.DataFrame(data)
df
```
