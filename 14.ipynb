Series

```python
#Creating series with default index

num = [1,2,3,4,5]

nw_series = pd.Series(num)

print(nw_series)
```

```python
#Creating series with custom index

nw_series = pd.Series([1,2,3,4,5], index=[1,2,3,4,5])

print(nw_series)
```

```python
fruits = ['apple', 'papol', 'kehel']

fr_series = pd.Series(fruits, index=[1,2,3])
print(fr_series)
```

```python
#Creating series from a dictionary
dct = {'name':['ksaun', 'kamal', 'nimal'],
       'country':'Finland',
       'city':'Helsinki'
       }

seris_fr_dicit = pd.Series(dct)
print(seris_fr_dicit)
```

```python
#Creating a pandas series using numpy linespace

nw_series = pd.Series(np.linspace(2,10, num=10, endpoint=False)) #endpoint decide whether to add the end number to the array

print(nw_series)
```

Modifing a Data Frame

```python
data = [
    {"Name": "Asabeneh", "Country":"Finland","City":"Helsinki"},
    {"Name": "David", "Country":"UK","City":"London"},
    {"Name": "John", "Country":"Sweden","City":"Stockholm"}
    ]

df = pd.DataFrame(data)
df
```

```python
#Adding an item (key/value pair) to a dataframe as pre defined list

weights = [78,74,80]
df['Weight'] = weights
print(df)

df['Height'] = [128,158,146]
df

```

```python
#convert height into meters

df['Height'] = df['Height'] * 0.01
df
```

```python
def calculate_bmi():
    weights = df['Weight']
    heights = df['Height']
    bmi =[]

    for w,h in zip(weights, heights):
        b = w/(h*h)
        bmi.append(b)

    return bmi

bmi = calculate_bmi()

print(bmi)
```

```python
df['BMI'] = bmi
df
```

```python
df['BMI'] = round(df['BMI'], 1)
df
```

```python
# We can use slicing to get specified rows from the dataframe. All columns will be selected, we cant specify 
# which column to select in slicing. Whether the indexing was 0,1,2,.. or any other we can use the slicing
df[0:4:2]
```

```python
# df.loc[row_label, coloumn_label], also called selection by label. The row and columns wcan be selected by labels. First we 
# need to pointout which row/rows we need, we can use slicing for it, if we have custom idex like a,b,c.. we can 
# use 'a':'b', thats why it called call by value, likewise the index is 0,1,2.. we use 0:1, But in here 
# both 0 and 1  both is included not like in other slicing because loc treat 0 and 1 as lables not as positions. 
# Then we can specify which columns we need ['column_1',..,..,..] 

print(df.loc[:, ['A', 'C']])
print(df.loc[dates[2], ['A','B']]) 
#In here dates[2] is the third element of dates list we created earlier. 
# So loc search by that value
print(df.loc[dates[2:4], ['A','B']]) 
```

```python
#More simple example --->

ndf = pd.DataFrame({
    'Name': ['John', 'Alice', 'Bob', 'David'],
    'Age': [25, 30, 22, 28],
    'Height': [175, 160, 180, 170]
}, index=['a', 'b', 'c', 'd'])
```

```python
print(ndf.loc[['a','c'], ['Name', 'Age']])
print(ndf.loc['a':'b', ['Name','Age', 'Height']])
```

```python
#Can use conditional filters

print(ndf.loc[ndf['Age'] > 25])
print(ndf.loc[ndf['Age'] > 25, ['Name']])
```
