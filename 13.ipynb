Concatanating two arrays. Adding is not possible because it will add values

```python
array_01 = np.array([[1,2],[3,4]], dtype=int)
array_02 = np.array([[10,20],[30,40]], dtype=int)

#Horizontal adding
#In horizontal number of rows of both arrays should be equal
print(np.hstack((array_01, array_02)))

#Vertical adding
#In vertical number of columns should be equal
print(np.vstack((array_01, array_02), dtype=float))
```

```python
print(np.concatenate((array_01,array_02), axis=0))

print('\n')

print(np.concatenate((array_01,array_02), axis=1))
```

Universal Functions -> ufunc

```python
# np.frompyfunc(function_name, number_of_input_arrays, number_of_output_arrays)


def add_num(a, b):
    return a+b

add_function = np.frompyfunc(add_num, 2, 1)

print(add_function([1,2,3,4], [10,20,30,40]))
```

```python
new_list =[]

def multiply_by_three(x):
    print("Received:", x)
    new_list.append(x)

my_ufunc = np.frompyfunc(multiply_by_three, 1, 0 )

arr = np.array([1, 2, 3, 4])

result = my_ufunc(arr)

print(result)
```

Summation

```python
arr_01 = np.array([[1,2,3,4], [5,6,7,8]])
arr_02 = np.array([[10,11,12,13], [14,15,16,17]])

#sum
print('Just sum:', np.sum([arr_01,arr_02]))

#can choose axis
#in here as u can see thre are two 2D arrays inside [] as parameters.So it become a 3D array. So axis changes
#But in a two 1D arrays inputed we get a one 2D array
print('Horizontal:', np.sum([arr_01,arr_02], axis=1))
print('Vertical:', np.sum([arr_01,arr_02], axis=0))
```

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([1, 2, 3])

print('Horizontal: ', np.sum([arr1, arr2], axis=1))
print('Vertical: ', np.sum([arr1, arr2], axis=0))


#Cumalative sum -> cumsum()
arr = np.array([1, 2, 3])

print(np.cumsum(arr))
```
